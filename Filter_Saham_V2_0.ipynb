```python
# ============================================================
# 3. TAHAP 1: SCREENING ANOMALI VOLUME & PRICE ACTION
# ============================================================
saham_lolos = []
print("Sedang melakukan screening anomali saham...")

for ticker in tickers:
    try:
        data = yf.download(ticker, start=start_date, end=end_date, progress=False)
        if data.empty:
            continue

        # Flatten MultiIndex jika ada (antisipasi update yfinance)
        if isinstance(data.columns, pd.MultiIndex):
            data.columns = data.columns.get_level_values(0)

        df_single = data.dropna()
        if len(df_single) < 21:
            continue

        last_row = df_single.iloc[-1]

        # --- Perhitungan Indikator Dasar ---
        vol_sma20 = df_single['Volume'].iloc[-21:-1].mean()
        rvol = last_row['Volume'] / vol_sma20 if vol_sma20 > 0 else 0
        turnover_val = last_row['Volume'] * last_row['Close']
        price_change_pct = ((last_row['Close'] - last_row['Open']) / last_row['Open']) * 100

        # --- [BARU] Perhitungan Indikator Tambahan ---
        # 1. MA20 untuk filter tren
        ma20 = df_single['Close'].rolling(window=20).mean().iloc[-1]

        # 2. Candle Strength (Kualitas Candle)
        candle_range = last_row['High'] - last_row['Low']
        candle_strength = (last_row['Close'] - last_row['Low']) / candle_range if candle_range > 0 else 0

        # --- SYarat Anomali (Diperketat) ---
        cond_rvol = rvol >= 2.5
        cond_turnover = turnover_val >= 2_000_000_000
        cond_bullish = last_row['Close'] > last_row['Open']
        cond_momentum = price_change_pct >= 2.5
        cond_trend = last_row['Close'] > ma20          # [BARU] Harga di atas MA20
        cond_candle = candle_strength >= 0.65          # [BARU] Close di 65% teratas range harian

        if cond_rvol and cond_turnover and cond_bullish and cond_momentum and cond_trend and cond_candle:
            saham_lolos.append(ticker)

    except Exception:
        continue

print(f"-> Selesai! Saham kandidat yang lolos screening ketat: {saham_lolos}\n")
```

```python

# ============================================================
# 4. TAHAP 2: DEEP ANALYSIS & TRADING PLAN (DENGAN ATR)
# ============================================================
if saham_lolos:
    hari_dict = {'Monday':'Senin', 'Tuesday':'Selasa', 'Wednesday':'Rabu',
                 'Thursday':'Kamis', 'Friday':'Jumat', 'Saturday':'Sabtu', 'Sunday':'Minggu'}

    for ticker in saham_lolos:
        try:
            data = yf.download(ticker, start=start_date, end=end_date, progress=False)
            if isinstance(data.columns, pd.MultiIndex):
                data.columns = data.columns.get_level_values(0)

            df_single = data.dropna()
            df_filtered = df_single[df_single.index <= tanggal_target]

            if df_filtered.empty:
                continue

            last_date = df_filtered.index[-1]
            last_row = df_filtered.iloc[-1]
            close_today = last_row['Close']
            next_bday = last_date + pd.tseries.offsets.BDay(1)

            # --- [BARU] Perhitungan ATR (Average True Range) 14 Hari ---
            high = df_filtered['High']
            low = df_filtered['Low']
            close = df_filtered['Close']
            prev_close = close.shift(1)

            tr1 = high - low
            tr2 = abs(high - prev_close)
            tr3 = abs(low - prev_close)
            tr = pd.concat([tr1, tr2, tr3], axis=1).max(axis=1)
            atr_14 = tr.rolling(window=14).mean().iloc[-1]

            # --- Manajemen Risiko Dinamis ---
            # Cut loss = 1.5 x ATR di bawah harga close
            sl_price_dynamic = close_today - (1.5 * atr_14)
            # Batasi maksimal cut loss di -10% untuk menjaga risk per trade
            sl_price = max(sl_price_dynamic, close_today * 0.90)

            trailing_trigger = close_today * 1.07
            trailing_lock = close_today * 1.05

            print(f"=================================================================")
            print(f" TRADING PLAN: {ticker}")
            print(f"=================================================================")
            print(f". Tanggal Sinyal       : {last_date.strftime('%Y-%m-%d')} ({hari_dict[last_date.day_name()]})")
            print(f". Target Eksekusi      : {next_bday.strftime('%Y-%m-%d')} ({hari_dict[next_bday.day_name()]})")
            print(f"-----------------------------------------------------------------")
            print(f". Harga Beli Acuan     : Rp{close_today:,.0f}")
            print(f". Cut Loss Dinamis     : Rp{sl_price:,.0f} (Berbasis ATR 14, Maks -10%)")
            print(f". Trailing Trigger     : Rp{trailing_trigger:,.0f} (+7%)")
            print(f". Trailing Lock        : Rp{trailing_lock:,.0f} (+5%)")
            print(f". Batas Hold Max       : 5 Hari Bursa")
            print(f"=================================================================\n")

        except Exception as e:
            print(f"Error processing {ticker}: {e}")
            continue

else:
    print("Hasil: Tidak ada saham yang memicu sinyal beli hari ini dengan kriteria ketat.")
```
